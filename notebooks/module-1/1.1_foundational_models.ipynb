## Customising your Model

```python
# model = init_chat_model(
#     model="gpt-5-nano",
#     # Kwargs passed to the model:
#     temperature=1.0
# )
model = ChatOllama(
    model="gemma4:e2b",
    # Kwargs passed to the model:

    # Temperature
    temperature=1.0,
    # Es el parámetro más conocido y controla el determinismo o la "creatividad".
    # Cómo funciona: Afecta a la distribución de probabilidad de las palabras. Una temperatura baja hace que el modelo siempre elija la palabra más probable.
    # Para qué sirve:
    # 0.0 - 0.3: Ideal para tareas precisas, extracción de datos, programación o responder preguntas basadas en hechos.
    # 0.7 - 1.0: Ideal para escritura creativa, lluvia de ideas o chatbots con personalidad.
    # > 1.0: El modelo empieza a volverse incoherente o "alucina" más de la cuenta.

    # Top-P (nucleus sampling)
    # top_p=,
    # En lugar de mirar todas las palabras posibles, el modelo solo considera un subconjunto cuya probabilidad sumada alcance el valor $P$.
    # Cómo funciona: Si $P = 0.9$, el modelo solo elegirá entre las palabras que juntas suman el 90% de la probabilidad total.
    # Para qué sirve: Ayuda a que el modelo no sea repetitivo pero mantenga la coherencia. Suele ser mejor dejar la Temperatura alta y controlar el caos con un Top-P de 0.8 o 0.9.

    # Top-K
    # top_k=,
    # Limita el vocabulario a un número fijo de palabras candidatas.
    # Cómo funciona: Si $K = 50$, el modelo solo considerará las 50 palabras más probables, sin importar qué tan probable sea la número 51.
    # Para qué sirve: Evita que el modelo elija palabras extremadamente raras o fuera de contexto (la "cola larga" de la distribución).

    # Penalización de Presencia y Frecuencia
    # repeat_penalty=,
    # Estos parámetros combaten uno de los problemas más comunes de los modelos locales: quedarse "trabados" repitiendo la misma frase.
    # Frequency Penalty: Penaliza palabras que ya han aparecido mucho en el texto. Usarlo si el modelo escribe "el, el, el" o usa mucho una muletilla.
    # Presence Penalty: Penaliza cualquier palabra que ya haya aparecido, forzando al modelo a hablar de temas nuevos. Usarlo para fomentar que el modelo sea más diverso en su vocabulario.

    # Ventana de Contexto (num_ctx o max_tokens)
    # num_ctx=,
    # max_tokens / max_output_tokens: Limita qué tan larga puede ser la respuesta del modelo. Útil para ahorrar memoria o evitar que el modelo se extienda demasiado.
    # num_ctx (en Ollama): Define cuánta memoria tiene el modelo para "recordar" el chat anterior o los documentos que le pasaste. En Gemma 4 puedes subir esto a 32000 o incluso más si tienes suficiente RAM.
)
response = model.invoke("What's the capital of the Moon?")
print(response.content)
```

```python
from langchain.messages import HumanMessage

response = agent.invoke(
    {"messages": [HumanMessage(content="What's the capital of the Moon?")]}
)
```

```python
from pprint import pprint

pprint(response)
```

```python
print(response['messages'][-1].content)
```

```python
from langchain.messages import AIMessage

response = agent.invoke(
    {"messages": [HumanMessage(content="What's the capital of the Moon?"),
    AIMessage(content="The capital of the Moon is Luna City."),
    HumanMessage(content="Interesting, tell me more about Luna City")]}
)

pprint(response)
```

## Streaming Output

```python
for token, metadata in agent.stream(
    {"messages": [HumanMessage(content="Tell me all about Luna City, the capital of the Moon")]},
    stream_mode="messages"
):

    # token is a message chunk with token content
    # metadata contains which node produced the token
    
    if token.content:  # Check if there's actual content
        print(token.content, end="", flush=True)  # Print token
```
